```python
import pandas as pd

df = pd.read_csv("baselines.csv")
df.head()
```

```python
table = df.groupby(["dataset", "model", "horizon"])["MAE"].agg(["mean", "std"])
table
```

```python
table = df.groupby(["dataset", "model", "horizon"])["RMSE"].agg(["mean", "std"])
table
```

```python
table = df.groupby(["dataset", "model", "horizon"])[["MAE", "RMSE"]].agg(["mean", "std"]).round(4)

table
```

```python
table.to_csv("mae_rmse_table.csv")
```
